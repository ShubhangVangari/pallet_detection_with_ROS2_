```python
from ultralytics import YOLO


model = YOLO("yolov8n.yaml")


data_yaml = "/content/drive/MyDrive/Projects/PR/Pallets_Detection/try_this/Pallet-Detection-2/data.yaml"



results = model.train(
    data=data_yaml,
    epochs=200,
    imgsz=640,
    batch=32,
    name="pallet_detector",
    lr0=0.01,
    lrf=0.001,
    momentum=0.937,
    weight_decay=0.0005,
    warmup_epochs=3,
    warmup_momentum=0.8,
    warmup_bias_lr=0.1,
    augment=True,
    cos_lr=True,
    freeze=None,
    optimizer="SGD",
    verbose=True
)


print(results)


model.save("/content/drive/MyDrive/Projects/PR/Pallets_Detection/models/TT_PD2_base_model_200_.pt")
```

```python
import os
import glob
from tqdm import tqdm

# -------------------------------
# 1. Define the Source Folders
# -------------------------------
images_folder = "/content/drive/MyDrive/Projects/PR/Pallets_Detection/try_this/Pallet-Detection-1/test/images"  # e.g., "/content/drive/MyDrive/Projects/PR/Pallets_Detection/dataset/pallet_detection-1/train/images"
labels_folder = "/content/drive/MyDrive/Projects/PR/Pallets_Detection/try_this/Pallet-Detection-1/test/labels"    # e.g., "/content/drive/MyDrive/Projects/PR/Pallets_Detection/dataset/pallet_detection-1/train/Labels_DINO"

# -------------------------------
# 2. Retrieve Image Paths
# -------------------------------
allowed_extensions = (".jpg", ".jpeg", ".png", ".bmp")
image_paths = sorted(glob.glob(os.path.join(images_folder, "*.*")))
# Filter image files to only allowed extensions (if needed)
image_paths = [p for p in image_paths if p.lower().endswith(allowed_extensions)]

# -------------------------------
# 3. Process Each Image-Label Pair with tqdm
# -------------------------------
for idx, img_path in enumerate(tqdm(image_paths, desc="Renaming image-label pairs"), start=1):
    # Get current image base name and extension
    old_base = os.path.splitext(os.path.basename(img_path))[0]
    ext = os.path.splitext(img_path)[1]

    # Construct new sequential base name (e.g. "000001")
    new_base = f"{idx:06d}"
    new_img_name = new_base + ext
    new_img_path = os.path.join(images_folder, new_img_name)

    # Rename image file
    os.rename(img_path, new_img_path)

    # Check if a corresponding label file exists and rename it accordingly.
    old_label_file = os.path.join(labels_folder, old_base + ".txt")
    if os.path.exists(old_label_file):
        new_label_file = os.path.join(labels_folder, new_base + ".txt")
        os.rename(old_label_file, new_label_file)
    else:
        # Optionally, handle missing label files (e.g., log a warning).
        pass

print("Renaming completed for all image and label pairs.")
```

```python
import os
import glob
import random
import shutil
from tqdm import tqdm

# -------------------------------
# 1. Define the Source and Destination Folders
# -------------------------------
# Training folders (current dataset)
train_images_folder = "/content/drive/MyDrive/Projects/PR/Pallets_Detection/try_this/Pallet-Detection-1/train/images"
train_labels_folder = "/content/drive/MyDrive/Projects/PR/Pallets_Detection/try_this/Pallet-Detection-1/train/labels"

# Validation folders (new location)
val_images_folder = "/content/drive/MyDrive/Projects/PR/Pallets_Detection/try_this/Pallet-Detection-1/valid/images"
val_labels_folder = "/content/drive/MyDrive/Projects/PR/Pallets_Detection/try_this/Pallet-Detection-1/valid/labels"

# Create validation folders if they don't exist.
os.makedirs(val_images_folder, exist_ok=True)
os.makedirs(val_labels_folder, exist_ok=True)

# -------------------------------
# 2. Retrieve File Paths and Build Pairs
# -------------------------------
allowed_ext = (".jpg", ".jpeg", ".png", ".bmp")
image_paths = sorted(glob.glob(os.path.join(train_images_folder, "*.*")))
image_paths = [p for p in image_paths if p.lower().endswith(allowed_ext)]

# Build a list of (image_path, label_path) pairs.
pairs = []
for img_path in image_paths:
    base_name = os.path.splitext(os.path.basename(img_path))[0]
    label_path = os.path.join(train_labels_folder, base_name + ".txt")
    if os.path.exists(label_path):
        pairs.append((img_path, label_path))
    # Optionally, you could log or handle images without a corresponding label.

# -------------------------------
# 3. Split the Data (15% for Validation)
# -------------------------------
VALID_SPLIT = 0.15  # 15% validation
random.seed(42)
random.shuffle(pairs)
val_count = int(len(pairs) * VALID_SPLIT)
val_pairs = pairs[:val_count]

# -------------------------------
# 4. Move the Selected Pairs to Validation Folders using tqdm
# -------------------------------
for img_path, label_path in tqdm(val_pairs, desc="Moving validation pairs"):
    # Move the image to the validation images folder.
    new_img_path = os.path.join(val_images_folder, os.path.basename(img_path))
    shutil.move(img_path, new_img_path)

    # Move the label file to the validation labels folder.
    new_label_path = os.path.join(val_labels_folder, os.path.basename(label_path))
    shutil.move(label_path, new_label_path)

print("Validation split created and corresponding image-label pairs moved.")
```
