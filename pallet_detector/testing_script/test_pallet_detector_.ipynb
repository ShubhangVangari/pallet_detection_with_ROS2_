```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

model = YOLO("/content/drive/MyDrive/Projects/PR/Pallets_Detection/models/TT_PD2_base_model_200_AMP_.pt")
image_path = "/content/drive/MyDrive/Projects/PR/Pallets_Detection/try_this/Pallet-Detection-1/test/images/000004.jpg"  

results = model.predict(source=image_path, imgsz=640, conf=0.25)  

annotated_frame = results[0].plot()  

plt.figure(figsize=(10, 10))
plt.imshow(cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title("Inference Output")
plt.show()
```

```python
import os
import glob
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt


input_folder = "/content/drive/MyDrive/Projects/PR/Pallets_Detection/try_this/Pallet-Detection-1/test/images"  

output_folder = "/content/drive/MyDrive/PR/Pallets_Detection/dataset/test/annotated_images_model_I_"  
os.makedirs(output_folder, exist_ok=True)

model = YOLO("/content/drive/MyDrive/Projects/PR/Pallets_Detection/models/TT_PD2_base_model_200_AMP_.pt")

image_paths = glob.glob(os.path.join(input_folder, '*.*'))



for image_path in image_paths:
    results = model.predict(source=image_path, imgsz=640, conf=0.25)
    annotated_frame = results[0].plot()
    base_filename = os.path.basename(image_path)

    plt.figure(figsize=(10, 10))
    plt.imshow(cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB))
    plt.title(f"Annotated: {base_filename}")
    plt.axis("off")
    plt.show()

```
