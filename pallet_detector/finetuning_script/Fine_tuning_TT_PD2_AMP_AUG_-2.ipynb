```python
model_path = '/content/drive/MyDrive/Projects/PR/Pallets_Detection/models/TT_PD2_base_model_200_.pt'
```

```python
from ultralytics import YOLO

model = YOLO(model_path)

model.train(
    data="/content/drive/MyDrive/Projects/PR/Pallets_Detection/try_this/Pallet-Detection-2/data.yaml",
    epochs=100,
    imgsz=640,
    batch=32,
    amp=True,
    lr0=0.001,
    lrf=0.01,
    momentum=0.937,
    weight_decay=0.0005,
    warmup_epochs=3,
    warmup_momentum=0.8,
    warmup_bias_lr=0.1,
    augment=True,
    cos_lr=True,
    freeze=None,
    optimizer="SGD",
    verbose=True
)

model.save("/content/drive/MyDrive/Projects/PR/Pallets_Detection/models/TT_PD2_base_model_200_AMP_.pt")
```
